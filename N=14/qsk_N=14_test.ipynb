# QSK (N=14)

```python
import sys
sys.path.append('../volume_law_utility')

from qsk_netket_pipeline import *
from sk import generate_hamiltonian_sk, generate_state_sk
from hrbm import *
```

```python
# 1. Setup SK Hamiltonian & Hilbert space
N = 14
seed = 1234
hi, H = generate_hamiltonian_sk(N=N, seed=seed)

# 2. Compute Exact Diagonalization (ED) target energy
_, ground_state_vec = generate_state_sk(N=N, seed=seed)
sp_h = H.to_sparse()
exact_E = (ground_state_vec.conj().T @ sp_h @ ground_state_vec).real
print(f"Exact ED Ground State Energy: {exact_E:.6f}")
```

# NQS

```python
# Common hyperparams
seed_nqs =101
key_model, key_sampler, key_vstate = set_seed(seed_nqs)

alpha=1
lr=0.05
n_samples = 1008
epochs=450
```

## RBM

```python
# 3. Define Sampler and Variational State for QSK
sampler = nk.sampler.MetropolisLocal(hi)

# Optimizer
optimizer = optax.inject_hyperparams(optax.sgd)(learning_rate=lr)

# Define real-valued RBM / HRBM model here
model = nk.models.RBM(alpha=1) 

# The variational state
vstate = nk.vqs.MCState(sampler, model, n_samples=n_samples, seed=key_vstate)
print("Has", vstate.n_parameters, "parameters")

start = time.time()
netket_pipeline(N, model, sampler, H, vstate, optimizer, 
            iter_num=epochs,  exact_E =exact_E,  holomorphic=False,
            log_filename=f"qsk_N={N}/seed={seed_nqs}/rbm_alpha={alpha}_ML")

end = time.time()
print(f'Time taken = {end-start} (secs)')
```

## HRBM

```python
# 3. Define Sampler and Variational State for QSK
sampler = nk.sampler.MetropolisLocal(hi)

# Optimizer
optimizer = optax.inject_hyperparams(optax.sgd)(learning_rate=lr)

# Define real-valued RBM / HRBM model here
k=1.0
Lmax=10
model_h = HyperbolicRBMd_Coupled(alpha=alpha, k=k, Lmax=Lmax)

# The variational state
vstate = nk.vqs.MCState(sampler, model_h, n_samples=n_samples, seed=key_vstate)
print("Has", vstate.n_parameters, "parameters")

start = time.time()
netket_pipeline(N, model_h, sampler, H, vstate, optimizer, 
            iter_num=epochs,  exact_E =exact_E,  holomorphic=False,
            log_filename=f"qsk_N={N}/seed={seed_nqs}/hrbm_alpha={alpha}_ML")

end = time.time()
print(f'Time taken = {end-start} (secs)')
```
