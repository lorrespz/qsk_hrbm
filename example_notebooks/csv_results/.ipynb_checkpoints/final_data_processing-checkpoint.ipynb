```python
def format_latex_sci(val, sig_figs=4):
    """
    Converts numbers into LaTeX scientific notation: 'X.XX \\times 10^{-Y}'
    Leaves non-numeric strings (like '--', 'HRBM') untouched.
    """
    try:
        num = float(val)
        if num == 0:
            return "$0$"
            
        # Format into scientific string e.g. "2.04e-04"
        formatted_sci = f"{num:.{sig_figs}e}"
        
        # Parse mantissa and exponent
        mantissa, exp = formatted_sci.split('e')
        exp_int = int(exp)
        
        if exp_int == 0:
            return f"${mantissa}$"
            
        return f"${mantissa} \\times 10^{{{exp_int}}}$"
    except (ValueError, TypeError):
        # Return string non-numeric entries as-is
        return str(val)
```

# Table presentation

### Overview

```python
def Nt(N, Nv):
    Nt = []
    for (k,j) in zip(N,Nv):
        Nt.append(k)
        Nt.append(j)
    return Nt
```

```python
df = pd.read_csv('scaling_benchmark_all_seeds_metrics.csv')
df
```

```python
df = pd.read_csv('scaling_benchmark_all_seeds_metrics.csv')
df.columns
```

### Energy table

```python
# ENERGY TABLE
df_energy = df[['N','model', 'mean_rel_err', 'sem_rel_err', 'mean_var', 'sem_var']]
#df_energy.fillna('--', inplace=True)
df_energy=df_energy.dropna()
df_energy
```

```python
# Energy table 
n_E = ['HRBM', '---', 'RBM', '---', 'HRBM', '---', 'RBM', '---','HRBM', '---', 'RBM', '---',
       'HRBM', '---', 'RBM', '---','HRBM', '---', 'RBM', '---', 'HRBM', '---', 'RBM', '---']
n_size = [14, '--', '--', '--', 16, '--', '--', '--', 18, '--', '--', '--', 20, '--', '--', '--', 22, '--', '--', '--',
        24, '--', '--', '--']
dd2 = {'N': n_size, 'model': n_E, 'mean_rel_err': Nt(df_energy.mean_rel_err, df_energy.sem_rel_err),
     'mean_var':Nt(df_energy.mean_var, df_energy.sem_var)}
df_e2=pd.DataFrame(dd2)
df_e2
```

```python
#print(df_e2.to_latex(index=False))
for col in ['mean_rel_err', 'mean_var']:
    df_e2[col] = df_e2[col].apply(format_latex_sci)

# Generate and print LaTeX code
latex_table = df_e2.to_latex(
    index=False,
    escape=False )  # Critical: allows LaTeX math symbols like $ and \ to render properly)

print(latex_table)
```
